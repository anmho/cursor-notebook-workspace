# Remote CUDA PyTorch workbook

Run this workbook only after opening the project in Cursor Remote SSH on a Linux NVIDIA host. It intentionally refuses CPU/MPS fallback.

```python
import platform

import torch

assert torch.cuda.is_available(), "This workbook requires an NVIDIA CUDA runtime."
device = torch.device("cuda")
print(
    {
        "platform": platform.platform(),
        "torch": torch.__version__,
        "cuda": torch.version.cuda,
        "device": torch.cuda.get_device_name(0),
    }
)
```

```python
left = torch.ones((1024, 1024), device=device)
right = torch.full((1024, 1024), 2.0, device=device)
product = left @ right
torch.cuda.synchronize()
print(
    {
        "device": str(product.device),
        "shape": tuple(product.shape),
        "sample": float(product[0, 0].cpu()),
    }
)
```

```python
x = torch.randn(128, 4, device=device, requires_grad=True)
weights = torch.randn(4, 1, device=device, requires_grad=True)
loss = ((x @ weights) ** 2).mean()
loss.backward()
torch.cuda.synchronize()
print({"loss": float(loss.detach().cpu()), "gradient_norm": float(weights.grad.norm().cpu())})
```
